# 01 Data Loading and Initial Inspection

This notebook loads the extracted performance dataset and performs initial summary statistics.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import json
import os

# Set plotting style
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('viridis')
```

## Load Datasets

```python
dataset_path = '../datasets/master_performance_dataset.csv'
df = pd.read_csv(dataset_path)
print(f"Loaded {len(df)} entries.")
df.head()
```

## Distribution by Dictionary

```python
plt.figure(figsize=(10, 6))
df['source_dictionary'].value_counts().plot(kind='barh')
plt.title('Entries per Dictionary')
plt.xlabel('Count')
plt.show()
```

## Distribution by Category

```python
plt.figure(figsize=(10, 6))
df['category'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Distribution of Categories')
plt.ylabel('')
plt.show()
```

## Top Subcategories

```python
plt.figure(figsize=(12, 8))
df[df['category'] != 'GENERAL']['subcategory'].value_counts().head(20).plot(kind='bar')
plt.title('Top 20 Performance-Related Subcategories')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.show()
```
