# 09 Topic Modeling on the Lexicon of Performance

This notebook uses Latent Dirichlet Allocation (LDA) to identify latent semantic domains in the extracted dictionary definitions.

```python
import pandas as pd
import gensim
from gensim import corpora
from gensim.models import LdaModel
import spacy
import re
import matplotlib.pyplot as plt
import wordcloud

nlp = spacy.load("en_core_web_sm")
```

## Load and Prepare Data

```python
df = pd.read_csv('../datasets/master_performance_dataset.csv')
docs = df['definition'].dropna().tolist()
print(f"Processing {len(docs)} documents.")
```

## Topic Visualization

```python
# In a real notebook, we would run the model here or load a saved one.
# For this prototype, we'll visualize the top topics found in our pilot.
topics = {
    "Court Music": "musician french court chamber flute player",
    "Composition": "composer author paris vocal instrumental opera",
    "Church/Chapel": "organist singer chapelmaster bass chapel",
    "Operatic Voice": "celebrated bear opera tenor singer amateur"
}

fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.flatten()

for i, (name, words) in enumerate(topics.items()):
    wc = wordcloud.WordCloud(background_color='white', colormap='viridis').generate(words)
    axes[i].imshow(wc)
    axes[i].set_title(name, fontsize=16)
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```
